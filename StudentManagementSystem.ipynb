```python
"""
Student Management System
"""

students = []


def add_student():
    """Collect details for a new student and add them to the list."""
    student_id = input("Enter Student ID: ").strip()
    name = input("Enter Name: ").strip()
    age = input("Enter Age: ").strip()
    course = input("Enter Course: ").strip()
    marks = input("Enter Marks: ").strip()

    student = {
        "ID": student_id,
        "Name": name,
        "Age": age,
        "Course": course,
        "Marks": marks
    }
    students.append(student)
    print(f"\nStudent '{name}' added successfully.\n")


def view_students():
    """Display all student records currently stored."""
    if not students:
        print("\nNo student records found.\n")
        return

    print("\n--- All Student Records ---")
    for student in students:
        print(f"ID: {student['ID']} | Name: {student['Name']} | "
              f"Age: {student['Age']} | Course: {student['Course']} | Marks: {student['Marks']}")
    print()


def search_student():
    """Search for a student by ID or Name (case-insensitive)."""
    if not students:
        print("\nNo student records found.\n")
        return

    keyword = input("Enter Student ID or Name to search: ").strip().lower()
    found = False

    for student in students:
        if keyword == str(student['ID']).lower() or keyword == student['Name'].lower():
            print(f"\nFound: ID: {student['ID']} | Name: {student['Name']} | "
                  f"Age: {student['Age']} | Course: {student['Course']} | Marks: {student['Marks']}\n")
            found = True

    if not found:
        print("\nNo matching student found.\n")


def update_student():
    """Update an existing student's details by ID. Blank input keeps the old value."""
    if not students:
        print("\nNo student records found.\n")
        return

    student_id = input("Enter Student ID to update: ").strip()

    for student in students:
        if str(student['ID']) == student_id:
            print("Leave a field blank to keep it unchanged.")
            name = input(f"New Name [{student['Name']}]: ").strip()
            age = input(f"New Age [{student['Age']}]: ").strip()
            course = input(f"New Course [{student['Course']}]: ").strip()
            marks = input(f"New Marks [{student['Marks']}]: ").strip()

            if name:
                student['Name'] = name
            if age:
                student['Age'] = age
            if course:
                student['Course'] = course
            if marks:
                student['Marks'] = marks

            print("\nStudent record updated successfully.\n")
            return

    print("\nStudent ID not found.\n")


def delete_student():
    """Delete a student record by ID."""
    if not students:
        print("\nNo student records found.\n")
        return

    student_id = input("Enter Student ID to delete: ").strip()

    for student in students:
        if str(student['ID']) == student_id:
            students.remove(student)
            print("\nStudent record deleted successfully.\n")
            return

    print("\nStudent ID not found.\n")


def display_menu():
    """Print the main menu options."""
    print("===== Student Management System =====")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search Student (by ID or Name)")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Exit")


def main():
    """Run the menu loop until the user chooses to exit."""
    while True:
        display_menu()
        choice = input("Enter your choice (1-6): ").strip()

        if choice == '1':
            add_student()
        elif choice == '2':
            view_students()
        elif choice == '3':
            search_student()
        elif choice == '4':
            update_student()
        elif choice == '5':
            delete_student()
        elif choice == '6':
            print("Exiting the program. Goodbye!")
            break
        else:
            print("\nInvalid choice. Please select a number between 1 and 6.\n")


if __name__ == "__main__":
    main()
```
